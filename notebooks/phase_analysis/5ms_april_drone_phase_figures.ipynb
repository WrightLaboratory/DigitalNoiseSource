```python
def drone_angles_for_corr_range_simple(dish_location, drone_file, corr_data_directory,
                                       start_file=0, stop_file=-1,
                                       use_ctime=False, output='degrees',
                                       lat_offset_m=0.0,
                                       lon_offset_m=0.0,
                                       alt_offset_m=0.0):
    """
    Same as before, but allows applying offsets (in meters) to drone position:
        lat_offset_m → north/south shift
        lon_offset_m → east/west shift
        alt_offset_m → vertical shift
    """

    # -----------------------
    # Step 1: Load correlator timestamps
    # -----------------------
    all_files = sorted([f for f in os.listdir(corr_data_directory) if ".lock" not in f])
    filenames = all_files[start_file:stop_file]
    if len(filenames) == 0:
        raise ValueError("No correlator files found in the specified range.")

    t_list = []
    for fname in filenames:
        with h5py.File(os.path.join(corr_data_directory, fname), 'r') as fd:
            if use_ctime:
                t_file = fd['index_map']['time']['ctime'][:].astype(float)
            else:
                t_file = 1e-9 * fd['index_map']['time']['irigb_time'][:].astype(float)
            t_list.append(t_file)
    t_corr = np.concatenate(t_list)

    t_corr_sec = np.array(t_corr, dtype=float)

    # -----------------------
    # Step 2: Load drone CSV timestamps
    # -----------------------
    df = pd.read_csv(drone_file, low_memory=False)

    if 'datetime(utc)' in df.columns and 'time(millisecond)' in df.columns:
        t0 = pd.to_datetime(df['datetime(utc)'], utc=True).iloc[0]
        drone_times = np.array([
            (t0 + datetime.timedelta(milliseconds=x)).timestamp()
            for x in df['time(millisecond)']
        ])
    elif 'GPS:dateTimeStamp' in df.columns:
        drone_times = pd.to_datetime(
            df['GPS:dateTimeStamp'], utc=True
        ).map(lambda t: t.timestamp()).values
    else:
        raise KeyError("No recognizable time columns in drone file.")

    # -----------------------
    # Step 3: Mask drone data
    # -----------------------
    t_start_sec, t_stop_sec = t_corr_sec[0], t_corr_sec[-1]
    mask = (drone_times >= t_start_sec) & (drone_times <= t_stop_sec)
    if not mask.any():
        raise ValueError("No drone data falls within the correlator time range.")

    df_masked = df.loc[mask].copy()
    drone_times = drone_times[mask]

    # -----------------------
    # Step 4: Extract positions
    # -----------------------
    if 'RTKdata:Lat_P' in df_masked.columns:
        lat = df_masked['RTKdata:Lat_P'].values
        lon = df_masked['RTKdata:Lon_P'].values
        alt = df_masked['RTKdata:Hmsl_P'].values
    elif 'GPS(0):Lat' in df_masked.columns:
        lat = df_masked['GPS(0):Lat'].values
        lon = df_masked['GPS(0):Long'].values
        alt = df_masked['GPS(0):heightMSL'].values
    elif 'GPS:Lat' in df_masked.columns:
        lat = df_masked['GPS:Lat'].values
        lon = df_masked['GPS:Long'].values
        alt = df_masked['GPS:heightMSL'].values
    elif 'latitude' in df_masked.columns:
        lat = df_masked['latitude'].values
        lon = df_masked['longitude'].values
        if 'altitude(meters)' in df_masked.columns:
            alt = df_masked['altitude(meters)'].values
        elif 'altitude(feet)' in df_masked.columns:
            alt = df_masked['altitude(feet)'].values * 0.3048
        else:
            raise KeyError("No altitude column found in drone file.")
    else:
        raise KeyError("No recognizable latitude/longitude columns in drone file.")

    # Remove NaNs
    valid_mask = ~np.isnan(lat) & ~np.isnan(lon) & ~np.isnan(alt)
    lat, lon, alt = lat[valid_mask], lon[valid_mask], alt[valid_mask]
    drone_times = drone_times[valid_mask]

    # -----------------------
    # NEW STEP: Apply offsets (in meters)
    # -----------------------
    meters_to_lat_deg = 1 / 111320.0
    meters_to_lon_deg = 1 / (111320.0 * np.cos(np.deg2rad(lat)))

    lat = lat + lat_offset_m * meters_to_lat_deg
    lon = lon + lon_offset_m * meters_to_lon_deg
    alt = alt + alt_offset_m

    # -----------------------
    # Step 5: Convert to local Cartesian
    # -----------------------
    origin_lat, origin_lon, origin_alt = dish_location
    prime_origin = pygeodesy.EcefCartesian(
        latlonh0=origin_lat, lon0=origin_lon, height0=origin_alt
    )

    drone_xyz = np.zeros((len(lat), 3))
    for i in range(len(lat)):
        p = pygeodesy.ellipsoidalNvector.LatLon(
            lat[i], lon[i], height=alt[i]
        )
        drone_xyz[i] = prime_origin.forward(p).toVector()

    # -----------------------
    # Step 6: Compute angles
    # -----------------------
    drone_xyz_rel = drone_xyz

    horizontal_dist = np.abs(drone_xyz_rel[:, 1])
    vertical_dist   = np.abs(drone_xyz_rel[:, 2])

    drone_angles = np.arctan2(horizontal_dist, vertical_dist)

    # -----------------------
    # Step 7: Interpolate
    # -----------------------
    angles_interp = np.interp(t_corr_sec, drone_times, drone_angles)

    if output.lower() == 'degrees':
        angles_interp = np.degrees(angles_interp)

    # -----------------------
    # Step 8: Diagnostics (unchanged)
    # -----------------------
    horizontal_x_dist = np.abs(drone_xyz_rel[:, 0])
    drone_angles_errors = np.arctan2(horizontal_x_dist, vertical_dist)

    horizontal_x_interp = np.interp(t_corr_sec, drone_times, horizontal_x_dist)
    angles_interp_errors = np.interp(t_corr_sec, drone_times, drone_angles_errors)

    if output.lower() == 'degrees':
        angles_interp_errors = np.degrees(angles_interp_errors)

    print(
        "Horizontal Deviation from Straight Path (x-direction):\n"
        "  Metric              |   Degrees   |   Meters\n"
        "  --------------------+-------------+-----------\n"
        f"  Max deviation       | {np.max(angles_interp_errors):9.4f} | {np.max(horizontal_x_interp):9.4f}\n"
        f"  Mean deviation      | {np.mean(angles_interp_errors):9.4f} | {np.mean(horizontal_x_interp):9.4f}\n"
        f"  Std. deviation      | {np.std(angles_interp_errors):9.4f} | {np.std(horizontal_x_interp):9.4f}"
    )

    return angles_interp
```

```python
import pygeodesy

def drone_distance_for_corr_range_simple(dish_location, drone_file, corr_data_directory,
                                         start_file=0, stop_file=-1,
                                         use_ctime=False,
                                         lat_offset_m=0.0,
                                         lon_offset_m=0.0,
                                         alt_offset_m=0.0):
    """
    Same as before, but allows applying offsets (in meters) to drone position:
        lat_offset_m → north/south shift
        lon_offset_m → east/west shift
        alt_offset_m → vertical shift
    """

    # -----------------------
    # Step 1: Load correlator timestamps
    # -----------------------
    all_files = sorted([f for f in os.listdir(corr_data_directory) if ".lock" not in f])
    filenames = all_files[start_file:stop_file]
    if len(filenames) == 0:
        raise ValueError("No correlator files found in the specified range.")

    t_list = []
    for fname in filenames:
        with h5py.File(os.path.join(corr_data_directory, fname), 'r') as fd:
            if use_ctime:
                t_file = fd['index_map']['time']['ctime'][:].astype(float)
            else:
                t_file = 1e-9 * fd['index_map']['time']['irigb_time'][:].astype(float)
            t_list.append(t_file)

    t_corr_sec = np.concatenate(t_list).astype(float)

    # -----------------------
    # Step 2: Load drone CSV timestamps
    # -----------------------
    df = pd.read_csv(drone_file, low_memory=False)

    if 'datetime(utc)' in df.columns and 'time(millisecond)' in df.columns:
        t0 = pd.to_datetime(df['datetime(utc)'], utc=True).iloc[0]
        drone_times = np.array([
            (t0 + datetime.timedelta(milliseconds=x)).timestamp()
            for x in df['time(millisecond)']
        ])
    elif 'GPS:dateTimeStamp' in df.columns:
        drone_times = pd.to_datetime(
            df['GPS:dateTimeStamp'], utc=True
        ).map(lambda t: t.timestamp()).values
    else:
        raise KeyError("No recognizable time columns in drone file.")

    # -----------------------
    # Step 3: Mask drone data
    # -----------------------
    t_start_sec, t_stop_sec = t_corr_sec[0], t_corr_sec[-1]
    mask = (drone_times >= t_start_sec) & (drone_times <= t_stop_sec)
    if not mask.any():
        raise ValueError("No drone data falls within the correlator time range.")

    df = df.loc[mask].copy()
    drone_times = drone_times[mask]

    # -----------------------
    # Step 4: Extract positions
    # -----------------------
    if 'RTKdata:Lat_P' in df.columns:
        lat = df['RTKdata:Lat_P'].values
        lon = df['RTKdata:Lon_P'].values
        alt = df['RTKdata:Hmsl_P'].values
    elif 'GPS(0):Lat' in df.columns:
        lat = df['GPS(0):Lat'].values
        lon = df['GPS(0):Long'].values
        alt = df['GPS(0):heightMSL'].values
    elif 'GPS:Lat' in df.columns:
        lat = df['GPS:Lat'].values
        lon = df['GPS:Long'].values
        alt = df['GPS:heightMSL'].values
    elif 'latitude' in df.columns:
        lat = df['latitude'].values
        lon = df['longitude'].values
        if 'altitude(meters)' in df.columns:
            alt = df['altitude(meters)'].values
        elif 'altitude(feet)' in df.columns:
            alt = df['altitude(feet)'].values * 0.3048
        else:
            raise KeyError("No altitude column found in drone file.")
    else:
        raise KeyError("No recognizable latitude/longitude columns in drone file.")

    # Remove NaNs
    valid = ~np.isnan(lat) & ~np.isnan(lon) & ~np.isnan(alt)
    lat, lon, alt = lat[valid], lon[valid], alt[valid]
    drone_times = drone_times[valid]

    # -----------------------
    # NEW STEP: Apply offsets (in meters)
    # -----------------------
    meters_to_lat_deg = 1 / 111320.0
    meters_to_lon_deg = 1 / (111320.0 * np.cos(np.deg2rad(lat)))

    lat = lat + lat_offset_m * meters_to_lat_deg
    lon = lon + lon_offset_m * meters_to_lon_deg
    alt = alt + alt_offset_m

    # -----------------------
    # Step 5: Convert to local Cartesian
    # -----------------------
    origin_lat, origin_lon, origin_alt = dish_location
    prime_origin = pygeodesy.EcefCartesian(
        latlonh0=origin_lat, lon0=origin_lon, height0=origin_alt
    )

    drone_xyz = np.zeros((len(lat), 3))
    for i in range(len(lat)):
        p = pygeodesy.ellipsoidalNvector.LatLon(
            lat[i], lon[i], height=alt[i]
        )
        drone_xyz[i] = prime_origin.forward(p).toVector()

    # -----------------------
    # Step 6: Distance
    # -----------------------
    y = drone_xyz[:, 1]
    z = drone_xyz[:, 2]

    drone_distance = np.sqrt(y**2 + z**2)

    # -----------------------
    # Step 7: Interpolate
    # -----------------------
    distance_interp = np.interp(t_corr_sec, drone_times, drone_distance)

    # -----------------------
    # Step 8: Diagnostics
    # -----------------------
    print(
        "Drone Distance from Dish (meters):\n"
        f"  Min: {distance_interp.min():.3f}\n"
        f"  Max: {distance_interp.max():.3f}\n"
        f"  Mean: {distance_interp.mean():.3f}\n"
        f"  Std: {distance_interp.std():.3f}"
    )

    return distance_interp
```

```python
import hdf5plugin

# ============================================================
# Inputs (match the script: directory of corr files, directory of ONE gain file)
# ============================================================
Data_Directory = '/hirax/DNS_freespace/yale_visit_april_2025/20250412T194037Z_WVUc52/corr/20250412T194134Z/'       
Gain_Directory = '/hirax/DNS_freespace/yale_visit_april_2025/20250412T191307Z_WVUc52_digitalgain/'  

Data_File_Index = [128, 131]        # start/stop slice like the script
fbounds = [0, 1024]              # [flb, fub]
CHIME_STYLE = False              # set True ONLY if you want the script's CHIME transpose behavior

# ============================================================
# Helpers: make prod iterable as (a,b) pairs, exactly as the script assumes
# ============================================================
def prod_iterable(prod_raw):
    """
    Return something you can do: for ii, pp in enumerate(prod): pp[0], pp[1]
    This is purely a compatibility shim for HDF5 dtypes / 1D encodings.
    """
    arr = np.asarray(prod_raw)

    # If it's a structured dtype (common in index_map tables), convert to Nx2
    if arr.dtype.fields is not None:
        fields = list(arr.dtype.fields.keys())
        if len(fields) < 2:
            raise ValueError(f"prod structured dtype has <2 fields: {fields}")
        return np.column_stack([arr[fields[0]], arr[fields[1]]]).astype(int)

    # If it's already Nx2, keep it
    if arr.ndim == 2 and arr.shape[1] == 2:
        return arr.astype(int)

    # If it's 1D where each element behaves like a length-2 tuple, just return as-is
    # (enumerate(prod_raw) will yield pp with pp[0], pp[1])
    return prod_raw

# ============================================================
# Build file list exactly like the script
# ============================================================
filelist = np.sort([x for x in os.listdir(Data_Directory) if ".lock" not in x])[:-1]
filenames = filelist[Data_File_Index[0] : Data_File_Index[1]]
n_files = len(filenames)

if n_files == 0:
    raise RuntimeError("No correlator files selected. Check Data_Directory and Data_File_Index.")

flb, fub = fbounds

print("Using correlator dir:", Data_Directory)
print("Selected files:", filenames[0], " ... ", filenames[-1])
print("Using gain dir:", Gain_Directory)

# ============================================================
# Load first data file for freq + prod shape context (like the script)
# ============================================================
with h5py.File(os.path.join(Data_Directory, filenames[0]), "r") as fd0:
    vis0 = fd0["vis"][:, flb:fub, :]
    if CHIME_STYLE:
        vis0 = np.array(fd0["vis"]).transpose(2, 0, 1)[:, flb:fub, :]
    freq = np.array([x[0] for x in fd0["index_map"]["freq"][flb:fub]])
    prod0 = fd0["index_map"]["prod"][:]

# ============================================================
# Load gain file EXACTLY like the script does
# ============================================================
gainfile = os.listdir(Gain_Directory)[0]
with h5py.File(os.path.join(Gain_Directory, gainfile), "r") as fg:
    digital_gain = fg["gain_coeff"][0]                 # <-- exactly script
    digital_gain *= np.power(2, fg["gain_exp"][0])[np.newaxis, :]  # <-- exactly script

gain = digital_gain.real[flb:fub, :]  # <-- exactly script

print("gainfile:", gainfile)
print("gain shape (freq,chan):", gain.shape)

# ============================================================
# Loop over corr files, apply "gain calibrate visibilities" exactly like script,
# and concatenate across files.
# ============================================================
vis_cal_list = []
t_list = []

for i, fname in enumerate(filenames):
    print(f"\rLoading + calibrating {i+1}/{n_files}: {fname}", end="")
    with h5py.File(os.path.join(Data_Directory, fname), "r") as fd:
        vis = fd["vis"][:, flb:fub, :]
        if CHIME_STYLE:
            vis = np.array(fd["vis"]).transpose(2, 0, 1)[:, flb:fub, :]

        # force complex-safe division (doesn't change values if already complex)
        vis = np.asarray(vis).astype(np.complex128, copy=False)

        prod = prod_iterable(fd["index_map"]["prod"][:])

        # --- THIS IS THE SCRIPT'S CALIBRATION LOGIC ---
        for ii, pp in enumerate(prod):
            vis[:, :, ii] /= (gain[:, pp[0]] * gain[:, pp[1]])[np.newaxis, :]

        # keep time axis simple here (script has ctime/irigb branches; add if you want)
        # We'll mimic the "irigb_time axis" branch since use_ctime defaults False in the script.
        tm = np.array(fd["index_map"]["time"]["irigb_time"])

    vis_cal_list.append(vis)
    t_list.append(tm)

print("\nDone. Concatenating...")

vis_cal_all = np.concatenate(vis_cal_list, axis=0)  # (time_total, n_freq, n_prod)
t_all = np.concatenate(t_list, axis=0)

print("vis_cal_all shape:", vis_cal_all.shape)
print("t_all shape      :", t_all.shape)
print("freq shape       :", freq.shape)
```

```python
def unwrap(phase_data):
    # Convert phase data to radians and unwrap
    unwrapped_data = np.unwrap(np.radians(phase_data))
    
    # Convert back to degrees
    unwrapped_data = np.degrees(unwrapped_data)
    
    # Align unwrapped data with the first point of the wrapped data
    adjustment = phase_data[-1] - unwrapped_data[-1]
    unwrapped_data += adjustment
    
    return unwrapped_data
```
